```python
# List Operations

# Creating a list

a = [10, 20, 30, 40, 50]

print(a)

# Adding an element

a.append(60)

print(a)

# Inserting an element

a.insert(2, 25)

print(a)

# Removing an element

a.remove(30)

print(a)

# Removing the last element

a.pop()

print(a)

```

```python
# Tuple Operations

# Creating a tuple

a = (10, 20, 30, 40, 50)

print(a)

# Finding length of tuple

print(len(a))

# Finding index of an element

print(a.index(30))

# Counting an element

print(a.count(20))

# Accessing an element

print(a[2])

```

```python
# Creating a dictionary

student = {
    "name": "Poorvi",
    "age": 18,
    "course": "CSE"
}

print(student)

print(student)
# Adding an element

student["city"] = "Pune"

print(student)

# Updating an element

student["age"] = 19

print(student)

# Removing an element

student.pop("city")

print(student)


# Accessing an element

print(student["name"])
```
